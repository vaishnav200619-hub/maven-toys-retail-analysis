# 02 Data Cleaning

This notebook cleans the raw files while preserving the original dataset.

```
from pathlib import Path
import pandas as pd
root = Path('C:/Users/gopak/maven-toys-retail-analysis/data/raw')
processed = Path('C:/Users/gopak/maven-toys-retail-analysis/data/processed')
processed.mkdir(exist_ok=True, parents=True)

def clean_currency(series):
    return series.astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False).str.strip().astype(float)

sales = pd.read_csv(root/'sales.csv', encoding='utf-8-sig')
products = pd.read_csv(root/'products.csv', encoding='utf-8-sig')
stores = pd.read_csv(root/'stores.csv', encoding='utf-8-sig')
inventory = pd.read_csv(root/'inventory.csv', encoding='utf-8-sig')

products['Product_Cost'] = clean_currency(products['Product_Cost'])
products['Product_Price'] = clean_currency(products['Product_Price'])
for df in [sales, stores]:
    for col in ['Date', 'Store_Open_Date']:
        if col in df.columns:
            df[col] = pd.to_datetime(df[col], errors='coerce')

for name, df in [('sales', sales), ('products', products), ('stores', stores), ('inventory', inventory)]:
    df = df.drop_duplicates()
    df.to_csv(processed / f'{name}_clean.csv', index=False)
    print(name, df.shape)
```

### Interpretation
The raw files are cleaned without modifying the original data. Currency conversion and datetime conversion are essential to support accurate revenue, cost, and trend analysis.
