# 01 Data Understanding

This notebook inspects the real Maven Toys tables and documents schema, quality, and relationships.

```
from pathlib import Path
import pandas as pd
root = Path('C:/Users/gopak/maven-toys-retail-analysis/data/raw')
files = sorted(root.glob('*.csv'))
for f in files:
    df = pd.read_csv(f, encoding='utf-8-sig')
    print(f'FILE: {f.name}')
    print('shape:', df.shape)
    print('columns:', list(df.columns))
    print('dtypes:\n', df.dtypes.to_string())
    print('missing_values:\n', df.isna().sum().to_string())
    print('duplicate_rows:', int(df.duplicated().sum()))
    print('head:\n', df.head(3).to_string(index=False))
    print('---')
```

### Interpretation
These tables are the actual source of truth for the product, sales, inventory, and store dimensions. The rows and columns confirm the business model before any joins or measures are created.
